```python
import hashlib, urllib.request
from google.colab import files, runtime
RUNNER_URL = "https://raw.githubusercontent.com/lluiseriksson/THE-ERIKSSON-PROGRAMME/48a2b9d0dbb87db864bd8fc4c879841ae8516240/scripts/colab_p0_p9_p4a_debug_validation.py"
RUNNER_SHA256 = "f20beb84b6998944d11c80889db130e990f847f5968e73290a0c65f572b4836f"
with urllib.request.urlopen(RUNNER_URL) as response:
    runner_source = response.read()
runner_hash = hashlib.sha256(runner_source).hexdigest()
print("RUNNER_TRANSPORT_SHA256=" + runner_hash, flush=True)
if runner_hash != RUNNER_SHA256:
    raise RuntimeError("RUNNER_TRANSPORT_HASH_MISMATCH")
release_runtime = runtime.unassign
runtime.unassign = lambda: print("RUNTIME_UNASSIGN_DEFERRED=1", flush=True)
launcher_exit = 0
try:
    exec(compile(runner_source, RUNNER_URL, "exec"), {"__name__": "__main__"})
except SystemExit as exc:
    launcher_exit = int(exc.code or 0)
except BaseException as exc:
    launcher_exit = 1
    print("LAUNCHER_EXCEPTION=" + repr(exc), flush=True)
finally:
    archive = "/content/hrpoly-p0-p9-p4a-debug-evidence.tar.gz"
    try:
        files.download(archive)
        print("EVIDENCE_DOWNLOAD_REQUESTED=1", flush=True)
    except Exception as exc:
        print("EVIDENCE_DOWNLOAD_ERROR=" + repr(exc), flush=True)
    print(f"LAUNCHER_EXIT={launcher_exit}", flush=True)
    release_runtime()
if launcher_exit != 0:
    raise RuntimeError(f"RUNNER_EXIT={launcher_exit}")
```
