```python
import hashlib, runpy, urllib.request
from pathlib import Path
RUNNER_URL = "https://raw.githubusercontent.com/lluiseriksson/THE-ERIKSSON-PROGRAMME/e33a91bdfb5a4e066e1df4aaf4673424111afac5/scripts/colab_cmp99_full_point_source_orientation_cold.py"
RUNNER_SHA256 = "608ee112c08c579b9fdf89a2e6bc36d4c4d91e0c68d70449e1d676b67c5494d6"
RUNNER_PATH = Path("/content/colab_cmp99_full_point_source_orientation_cold_v3.py")
with urllib.request.urlopen(RUNNER_URL, timeout=60) as response:
    runner_source = response.read()
runner_hash = hashlib.sha256(runner_source).hexdigest()
print("COLD_V3_RUNNER_TRANSPORT_SHA256=" + runner_hash, flush=True)
if runner_hash != RUNNER_SHA256:
    raise RuntimeError("COLD_V3_RUNNER_TRANSPORT_HASH_MISMATCH")
RUNNER_PATH.write_bytes(runner_source)
try:
    runpy.run_path(str(RUNNER_PATH), run_name="__main__")
except SystemExit as exit_status:
    exit_code = 0 if exit_status.code is None else int(exit_status.code)
    if exit_code != 0:
        raise
```
