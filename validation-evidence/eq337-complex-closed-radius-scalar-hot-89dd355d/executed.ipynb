```python
import hashlib, urllib.request
from google.colab import runtime
RUNNER_URL = "https://raw.githubusercontent.com/lluiseriksson/THE-ERIKSSON-PROGRAMME/c6272ab7c8fcbbefc1f5fdc075c976c1165d182f/scripts/colab_eq337_complex_closed_radius_scalar_validation.py"
RUNNER_SHA256 = "2b4a0e2b8443cc7f9cfe96689045cf9f1b09df4fbd833e691ceda959d87f6094"
with urllib.request.urlopen(RUNNER_URL) as response:
    runner_source = response.read()
measured = hashlib.sha256(runner_source).hexdigest()
print("RUNNER_TRANSPORT_SHA256=" + measured, flush=True)
if measured != RUNNER_SHA256:
    raise RuntimeError("RUNNER_TRANSPORT_HASH_MISMATCH")
release_runtime = runtime.unassign
runtime.unassign = lambda: print("RUNTIME_UNASSIGN_DEFERRED=1", flush=True)
launcher_exit = 0
try:
    exec(compile(runner_source, RUNNER_URL, "exec"), {"__name__": "__main__"})
except SystemExit as exc:
    launcher_exit = int(exc.code or 0)
except BaseException as exc:
    launcher_exit = 1
    print("LAUNCHER_EXCEPTION=" + repr(exc), flush=True)
finally:
    print(f"LAUNCHER_EXIT={launcher_exit}", flush=True)
    if launcher_exit == 0:
        release_runtime()
    else:
        print("RUNTIME_RETAINED_FOR_DEBUG=1", flush=True)
if launcher_exit != 0:
    raise RuntimeError(f"RUNNER_EXIT={launcher_exit}")
```

```python
import hashlib, pathlib, subprocess, urllib.request, time
SOURCE_SHA = "7f2a7cf16fa3493bb178833af51be3861fda5c4b"
SOURCE_URL = "https://raw.githubusercontent.com/lluiseriksson/THE-ERIKSSON-PROGRAMME/" + SOURCE_SHA + "/tmp/CMP99ComplexClosedRadiusScalar.repro.lean"
EXPECTED_SHA256 = "0d0aa5249feb205eb1b353b5eb16df590e8116a95a191294c53f5145c55f8857"
TARGET = pathlib.Path("/content/hrpoly-complex-closed-radius-scalar/tmp/CMP99ComplexClosedRadiusScalar.repro.lean")
data = urllib.request.urlopen(SOURCE_URL).read()
measured = hashlib.sha256(data).hexdigest()
print("HOT_SOURCE_SHA=" + SOURCE_SHA, flush=True)
print("HOT_SOURCE_SHA256=" + measured, flush=True)
assert measured == EXPECTED_SHA256, "HOT_SOURCE_HASH_MISMATCH"
TARGET.write_bytes(data)
cmd = ["lake","env","lean","tmp/CMP99ComplexClosedRadiusScalar.repro.lean","-o",".lake/build/lib/lean/tmp/CMP99ComplexClosedRadiusScalar.repro.olean"]
t0=time.perf_counter()
r=subprocess.run(cmd,cwd="/content/hrpoly-complex-closed-radius-scalar",stdout=subprocess.PIPE,stderr=subprocess.STDOUT,text=True)
print(r.stdout, end="", flush=True)
print("HOT_SCALAR_EXIT=" + str(r.returncode), flush=True)
print("HOT_SCALAR_SECONDS=" + f"{time.perf_counter()-t0:.3f}", flush=True)
```
