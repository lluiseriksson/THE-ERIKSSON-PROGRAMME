```python
import hashlib, urllib.request
URL = "https://raw.githubusercontent.com/lluiseriksson/THE-ERIKSSON-PROGRAMME/3ac847fdbaeb831a695503a7c1983f222755a220/scripts/colab_cmp89_eq248_source_role_cold.py"
EXPECTED = "f0a6e4b2007c23c07ac840b73cc74fdf3c6f1d8100c87ba2b4a6928fc1aaeb89"
data = urllib.request.urlopen(URL, timeout=60).read()
measured = hashlib.sha256(data).hexdigest()
print("COLD_RUNNER_SHA256=" + measured, flush=True)
if measured != EXPECTED:
    raise RuntimeError("COLD_RUNNER_HASH_MISMATCH")
exec(compile(data, URL, "exec"))
```
