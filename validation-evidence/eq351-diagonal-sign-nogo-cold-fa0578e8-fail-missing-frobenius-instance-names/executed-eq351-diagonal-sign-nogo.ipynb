# CMP99 Eq. (3.51) diagonal-sign no-go — cold validation

This notebook fetches one runner from an exact Git checkpoint, verifies its SHA-256, executes it once in a fresh Colab CPU/high-RAM runtime, and downloads the resulting evidence archive.

```python
import hashlib, pathlib, urllib.request
from google.colab import files

RUNNER_CHECKPOINT = 'c737f32a2d71a3bb3d86769cb55bb5eeb001f506'
RUNNER_SHA256 = 'ef395bb2c9ad068c809971eff43885ecfcba0c8a7de2dc3ae8eb2ab3ec8ae4f4'
RUNNER_URL = (
    'https://raw.githubusercontent.com/lluiseriksson/'
    'THE-ERIKSSON-PROGRAMME/' + RUNNER_CHECKPOINT +
    '/scripts/colab_cmp99_eq351_diagonal_sign_nogo_validation.py'
)
mem_kib = int(pathlib.Path('/proc/meminfo').read_text().splitlines()[0].split()[1])
print(f'RUNNER_CHECKPOINT={RUNNER_CHECKPOINT}', flush=True)
print(f'MEMTOTAL_KIB={mem_kib}', flush=True)
if mem_kib < 40_000_000:
    raise RuntimeError('HIGH_RAM_REQUIRED: expected at least 40,000,000 KiB')
runner_bytes = urllib.request.urlopen(RUNNER_URL, timeout=60).read()
actual = hashlib.sha256(runner_bytes).hexdigest()
print(f'RUNNER_SHA256={actual}', flush=True)
if actual != RUNNER_SHA256:
    raise RuntimeError(f'RUNNER_HASH_MISMATCH expected={RUNNER_SHA256} actual={actual}')
namespace = {'__name__': 'cmp99_eq351_diagonal_sign_nogo_runner'}
exec(compile(runner_bytes, RUNNER_URL, 'exec'), namespace)
exit_code = None
try:
    namespace['main']()
except SystemExit as exc:
    exit_code = int(exc.code or 0)
archive = pathlib.Path('/content/hrpoly-cmp99-eq351-diagonal-sign-nogo-evidence.tar.gz')
if archive.exists():
    print(f'DOWNLOAD_SHA256={hashlib.sha256(archive.read_bytes()).hexdigest()}', flush=True)
    files.download(str(archive))
if exit_code != 0:
    raise RuntimeError(f'RUNNER_EXIT={exit_code}')
```
