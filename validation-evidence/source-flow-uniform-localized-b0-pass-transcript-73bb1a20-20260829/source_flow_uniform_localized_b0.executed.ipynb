```python
import urllib.request, hashlib
URL = 'https://raw.githubusercontent.com/lluiseriksson/THE-ERIKSSON-PROGRAMME/a754e28555270ebbdab8bff05f262368a39c43fa/scripts/colab_source_flow_uniform_point_source_b0_validation.py'
EXPECTED = '2563bfba3de16b09733d2ae751a7f39ed0d29282386857ef51f8ed0767a29021'
payload = urllib.request.urlopen(URL, timeout=60).read()
measured = hashlib.sha256(payload).hexdigest()
print('LAUNCHER_SHA256=' + measured, flush=True)
if measured != EXPECTED:
    raise RuntimeError('LAUNCHER_HASH_MISMATCH')
exec(compile(payload, URL, 'exec'), {'__name__': '__main__'})
```
